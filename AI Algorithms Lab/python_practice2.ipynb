```python
import math
def is_prime(n):
    for i in range(2, int(math.sqrt(n) + 1)):
        if n%i == 0:
            return False
        return True

print(is_prime(19))
```

```python
def factorial(n):
    if n==0:
        return 1
    return n*factorial(n-1)

print(factorial(6))
```

```python
def sum(lst):
    sum =0
    for num in lst:
        sum += num
    return sum

print(sum([2,5,1]))
        
```

```python
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade
    def get_name(self):
        return self.name
    def get_grade(self):
        return self.grade
    def set_name(self, newname):
        self.name = newname
    def set_grade(self, newgrade):
        self.grade = newgrade

s1 = Student('Sah' , 'A')
s1.set_name('Sahil')
s1.set_grade('B')
print(s1.get_name())
print(s1.get_grade())
```
