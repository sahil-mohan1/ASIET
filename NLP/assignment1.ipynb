```python
import nltk
import string
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
```

### NLTK Downloads

```python
# --- Required NLTK Downloads ---
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
nltk.download('punkt_tab')
```

### Self-Introduction Paragraph

```python
# --- The Self-Introduction Paragraph ---
self_introduction = """
Hello! My name is Sahil Mohan. I am from Kozhikode, Kerala. I am a final year Computer Science and Engineering student specializing in Artificial Intelligence at the Adi Shankara Institute of Engineering and Technology.

Coming from a supportive family of four people including my parents and sister, my current status revolves around being an avid student and AI engineer. My achievements include obtaining an elite certification in NPTEL Machine Learning and working on a real-world Edge AI water level monitoring system at ICFOSS.

My hobbies involve watching movies and travelling. My core technical interests lie in Artificial Intelligence, Machine Learning, and Data Science, with a particular focus on building intelligent systems and deploying edge AI solutions.

My strengths lie in quickly grasping new concepts, tools, and environments, allowing me to stay productive and deliver results under changing requirements. A weakness I am actively managing is spending too much time perfecting the finer details of a project rather than prioritizing the core delivery milestone.

Check out my work at https://github.com/sahil-mohan1 or email me at sahil10au@gmail.com!
"""
```

### Step 1: Sentence Tokenization

```python
# --- Step 1: Sentence Tokenization ---
# Extracting sentences directly from the raw paragraph
sentences = sent_tokenize(self_introduction)
```

### Step 2: Word Tokenization

```python
# --- Step 2: Word Tokenization ---
# Extracting words directly from the raw paragraph
raw_words = word_tokenize(self_introduction)
```

### Step 3: Lowercasing and Punctuation Removal

```python
# --- Step 3: Combined Lowercasing and Punctuation Removal ---
# We loop through the tokenized words, lowering them and dropping them if they are isolated punctuation marks
cleaned_words = [word.lower() for word in raw_words if word not in string.punctuation]
```

### Step 4: Stop Word Removal

```python
# --- Step 4: Stop Word Removal ---
stop_words_set = set(stopwords.words('english'))
filtered_words = [word for word in cleaned_words if word not in stop_words_set]
```

### Step 5: Stemming

```python
# --- Step 5: Stemming ---
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]
```

### Step 6: Lemmatization

```python
# --- Step 6: Lemmatization ---
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]
```

### Verification Outputs

```python
# --- Verification Outputs ---
print(f"Total Sentences Tokenized: {len(sentences)}\n")

print("1. Cleaned Words (Lowercased and Punctuation Removed):")
print(cleaned_words, "\n")

print("2. Words after Stop Word Removal (Full List):")
print(filtered_words, "\n")

print("3. Stemmed Words (Full List):")
print(stemmed_words, "\n")

print("4. Lemmatized Words (Full List):")
print(lemmatized_words)
```
